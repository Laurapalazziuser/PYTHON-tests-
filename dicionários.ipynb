4. Percorrendo Chaves e Valores (Intermediário)


```python
capitais = {
    'Brasília': 'Brasil',
    'Paris': 'França',
    'Tóquio': 'Japão',
    'Lisboa': 'Portugal'
}

for capital, pais in capitais.items():   #O método .items() é muito útil: ele retorna os pares de chave e valor do dicionário ao mesmo tempo.
#No for, usamos duas variáveis temporárias: capital (que vai receber a chave, ou seja, o nome da cidade) e pais (que vai receber o valor, ou seja, o nome do país).
#A cada volta (iteração) do laço, o Python passa para a próxima linha do dicionário automaticamente.
    print(f"A capital {capital} pertence ao país {pais}")
```

```python
capitais = {
    'Brasília': 'Brasil',
    'Paris': 'França',
    'Tóquio': 'Japão',
    'Lisboa': 'Portugal'
}
# Usando apenas .keys() para iterar e buscando o valor correspondente no dicionário (Esse método retorna apenas uma lista contendo as chaves do dicionário:)
for capital in capitais.keys():  #A cada rodada do laço, a variável capital recebe uma dessas chaves. Na primeira rodada, capital vale 'Brasília'.
    pais = capitais[capital]  #No Python, quando passamos uma chave dentro de colchetes [ ] após o nome do dicionário, ele busca o valor associado a ela.
     #Na primeira rodada, capitais['Brasília'] devolve o valor correspondente, que é 'Brasil'. Esse valor é guardado na variável pais.
    print(f"A capital {capital} pertence ao país {pais}")
```

5. Removendo Elementos com .pop() (Intermediário)


```python
produtos = {
    'lapis': 2.50,
    'caderno': 15.00,
    'borracha': 1.00,
    'caneta': 3.00
}

palavra_remover = input('Digite o nome do produto que deseja remover: ')

# Verificamos diretamente no dicionário de produtos
if palavra_remover in produtos:
    # O .pop() remove o produto do dicionário. Opcionalmente, guardamos o preço removido.
    preco_removido = produtos.pop(palavra_remover)  #Ele encontra e remove o par chave-valor: Quando você passa a chave (por exemplo, 'caderno') dentro de produtos.pop('caderno'), o Python localiza essa chave no dicionário, apaga a chave e o preço correspondente da memória, atualizando o dicionário instantaneamente.
#Ele te devolve (retorna) o valor removido: Ao mesmo tempo em que apaga o item do dicionário, o .pop() "segura" o valor associado àquela chave (no caso, o preço 15.00) e o entrega de volta para você.
#Por isso, quando escrevemos:
#preco_removido = produtos.pop(palavra_remover)
#O Python remove o produto de produtos (atualizando o dicionário) e joga o preço dele para dentro da variável preco_removido!
    print(f"Foi removido: {palavra_remover} (Preço: R$ {preco_removido:.2f})")
    print(f"O dicionário agora fica: {produtos}")
else:
    print(f"O produto '{palavra_remover}' não foi encontrado no dicionário.")
```

7. Contagem de Frequência de Caracteres (Avançado)

```python
texto = input('Digite sua string: ').lower() # Convertemos para minúsculas para não diferenciar 'A' de 'a'
contagem = {}

for caractere in texto:
    # Desconsideramos espaços em branco para focar apenas nas letras/caracteres
    if caractere != ' ':
        if caractere in contagem:
            contagem[caractere] += 1 # Se já existe, soma 1
        else:
            contagem[caractere] = 1  # Se não existe, cria a chave com valor 1

print("Frequência de caracteres:")
print(contagem)
```

8. Dicionários Aninhados - Sistema de Alunos (Avançado)

```python
escola = {
    'M01': {
        'nome': 'Alice',
        'idade': 17,
        'notas': [8.0, 9.0, 7.0]
    },
    'M02': {
        'nome': 'Bruno',
        'idade': 18,
        'notas': [6.0, 7.5, 8.0]
    },
    'M03': {
        'nome': 'Carla',
        'idade': 17,
        'notas': [9.5, 10.0, 9.0]
    }
}

# Percorrendo o dicionário principal para calcular e exibir a média
for matricula, info in escola.items():
    nome = info['nome']
    notas = info['notas']
    media = sum(notas) / len(notas)
    print(f"Aluno(a): {nome} (Matrícula: {matricula}) - Média: {media:.2f}")
```

9. Lista de Dicionários - Catálogo de Filmes (Avançado)

```python
filmes = [
    {
        'titulo': 'O Poderoso Chefão',
        'diretor': 'Francis Ford Coppola',
        'ano': 1972
    },
    {
        'titulo': 'Pulp Fiction',
        'diretor': 'Quentin Tarantino',
        'ano': 1994
    },
    {
        'titulo': 'Metrópolis',
        'diretor': 'Fritz Lang',
        'ano': 1927
    }
]

# Definimos o primeiro filme como o mais antigo inicialmente
filme_mais_antigo = filmes[0]

# Percorremos a lista para comparar os anos de lançamento
for filme in filmes:
    if filme['ano'] < filme_mais_antigo['ano']:
        filme_mais_antigo = filme

print(f"O filme mais antigo é: '{filme_mais_antigo['titulo']}' lançado em {filme_mais_antigo['ano']}.")
```
