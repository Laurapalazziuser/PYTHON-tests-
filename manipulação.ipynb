```python
# Exercício 1 - Criando o arquivo e escrevendo a mensagem com o gerenciador de contexto 'with'
with open('mensagem.txt', 'w', encoding='utf-8') as arquivo:    # with garante que o arquivo seja aberto e, o mais importante, fechado automaticamente
# W é o modo de abertura. Ele diz ao Python para criar o arquivo se ele não existir, ou sobrescrever tudo o que estiver dentro dele caso já exista.
# AS apelida o arquivo aberto de arquivo, permitindo que você interaja com ele
    arquivo.write('Estudando manipulação de arquivos em Python!')

print('Arquivo "mensagem.txt" criado com sucesso!')  #Se você não usar o with, aí sim você é obrigado a usar o .close() manualmente, caso contrário o arquivo fica preso na memória do computador:
```

3- Adicionando Novos Dados (Modo Append)


```python
# Exercício 3 - Adicionando novos dados usando o modo append ('a')
with open('mensagem.txt', 'a', encoding='utf-8') as arquivo:
    # Usamos \n para quebrar a linha antes de escrever o novo texto
    arquivo.write('\nEsta linha foi adicionada posteriormente.')

print("Nova linha adicionada com sucesso!")

# Lendo o arquivo inteiro para confirmar que as duas linhas estão lá
with open('mensagem.txt', 'r', encoding='utf-8') as arquivo:
    conteudo_completo = arquivo.read()

print("\nConteúdo atualizado do arquivo:")
print(conteudo_completo)
```

4- Listagem de Compras Linha por Linha

```python
# 1. Gravando cada item em uma nova linha usando \n
with open('compras.txt', 'w', encoding='utf-8') as arquivo:
    arquivo.write('Maçã\n')
    arquivo.write('Leite\n')
    arquivo.write('Pão\n')
    arquivo.write('Café\n')
print("Lista de itens guardada com sucesso!\n")

# 2. Lendo linha por linha e enumerando
print("Lista de Compras Numerada:")
with open('compras.txt', 'r', encoding='utf-8') as arquivo:
    # O enumerate(..., 1) começa a contagem a partir do número 1
    for indice, linha in enumerate(arquivo, 1):
        # .strip() remove o \n invisível do final de cada linha para evitar linhas em branco extras
        print(f"{indice}. {linha.strip()}")
```

```python
# Forma alternativa e mais simples sem usar 'enumerate'

print("Lista de Compras (Usando um contador simples):")

with open('compras.txt', 'r', encoding='utf-8') as arquivo:
    contador = 1  # Criamos uma variável começando em 1

    for linha in arquivo:  # Para cada linha dentro do arquivo
        print(f"{contador}. {linha.strip()}")
        contador = contador + 1  # Somamos +1 para o próximo item
```

5- Gravando Entradas do Usuário

```python
# 1. Recebendo os nomes dos amigos do usuário
nome1 = str(input("Digite nome do seu amigo 1: "))
nome2 = str(input("Digite nome do seu amigo 2: "))
nome3 = str(input("Digite nome do seu amigo 3: "))

# 2. Gravando os nomes no arquivo 'amigos.txt'
with open('amigos.txt', 'w', encoding='utf-8') as arquivo:
    arquivo.write(f"{nome1}\n")
    arquivo.write(f"{nome2}\n")
    arquivo.write(f"{nome3}\n")

print("\nNomes salvos com sucesso! Agora lendo o arquivo:\n")

# 3. Lendo e exibindo o conteúdo do arquivo
with open('amigos.txt', 'r', encoding='utf-8') as arquivo:
    arquivo_completo = arquivo.read()

print(arquivo_completo)
```

6- Contador de Linhas

```python
# Exercício 6 - Contador de Linhas
# Vamos abrir o arquivo 'compras.txt' que criamos no Exercício 4

with open('compras.txt', 'r', encoding='utf-8') as arquivo:
    # O .readlines() lê todo o arquivo e salva cada linha como um item em uma lista
    linhas = arquivo.readlines()

# O len() calcula a quantidade de itens que existem dentro dessa lista (ou seja, as linhas)
total_linhas = len(linhas)

print(f"O arquivo 'compras.txt' possui {total_linhas} linhas.")
```

7- Buscando uma Palavra-Chave

```python
# 1. Criando o arquivo com as notas
with open('notas.txt', 'w', encoding='utf-8') as arquivo:
    arquivo.write('Matemática: Aprovado\n')
    arquivo.write('História: Reprovado\n')
    arquivo.write('Física: Aprovado\n')
print("Lista criada com sucesso!\n")

# 2. Buscando a palavra-chave 'Aprovado' linha por linha
print("Matérias em que foi Aprovado:")
with open("notas.txt", "r", encoding="utf-8") as arquivo:
    for linha in arquivo:
        # Usamos o operador 'in' para checar se a palavra está na linha
        if 'Aprovado' in linha:
            print(linha.strip())
```

8- Calculando com Dados de Arquivo

```python
# 1. Gravando os números no arquivo
with open('numeros.txt', 'w', encoding='utf-8') as arquivo:
     arquivo.write('10\n')
     arquivo.write('20\n')
     arquivo.write('30\n')
     arquivo.write('40\n')
     arquivo.write('50\n')
print('Lista criada com sucesso!\n')

# 2. Calculando a soma e a média
soma = 0
quantidade = 0

with open('numeros.txt', 'r', encoding='utf-8') as arquivo:
     for linha in arquivo:
          # Convertemos o texto da linha em um número inteiro real
          numero = int(linha.strip())
          soma = soma + numero
          quantidade = quantidade + 1

# Calculamos a média fora do loop, depois de somar todos os valores
media = soma / quantidade

print(f"Soma total: {soma}")
print(f"Média dos valores: {media}")
```

9- Gerando um Relatório de estatísticas


```python
# 1. Ler o arquivo de números e calcular a soma e a média
soma = 0
quantidade = 0

with open('numeros.txt', 'r', encoding='utf-8') as arquivo:
    for linha in arquivo:
        numero = int(linha.strip())
        soma = soma + numero
        quantidade = quantidade + 1

media = soma / quantidade

# 2. Gravar os resultados calculados no novo arquivo 'relatorio.txt'
with open('relatorio.txt', 'w', encoding='utf-8') as arquivo_saida:
    arquivo_saida.write(f"Soma total: {soma}\n")
    arquivo_saida.write(f"Média: {media}\n")

print("Relatório gerado e salvo em 'relatorio.txt' com sucesso!")
```

10- Frequência de Vogais em um Texto

```python
# 1. Criando o arquivo de texto
with open('texto_livre.txt', 'w', encoding='utf-8') as arquivo:
     arquivo.write('Meu nome é Laura e tenho 18 anos, serei uma cientista')
print("Arquivo criado com sucesso\n")

# 2. Contando as vogais de verdade
lista_vogais = 'AEIOUaeiouÁÉÍÓÚáéíóúÂÊÎÔÛâêîôûÃÕãõ'
contador = 0

with open('texto_livre.txt', 'r', encoding='utf-8') as arquivo:
     for linha in arquivo:               # Passa por cada linha do arquivo
          for letra in linha:            # Passa por cada letra daquela linha
               if letra in lista_vogais: # Verifica se a letra é uma vogal
                    contador += 1

print(f'Total de vogais encontradas: {contador}')
```
