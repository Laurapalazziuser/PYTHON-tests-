  2. Verificação de Número Primo:

```python
numero= int(input("Digite um número natural maior que zero: "))
contador_divisor=0
for i in range (1,numero+1):
  if numero % i == 0:
   contador_divisor +=1
# The primality check should happen after the loop has finished counting all divisors
  if contador_divisor == 2:
   print("o número é primo")
  else:
   print("não é primo")
```

2. Jogo da Adivinhação com Dicas por Faixa

```python
secret_number = 42 # Define um número secreto fixo
palpites = [10, 70, 50, 40, 45, 42, 1, 100] # Lista de palpites do jogador fictício

tentativas = 0
acertou = False

# Define um limite máximo para a diferença, usado para escalar a barra de progresso.
# Por exemplo, se o número secreto é 42 e os palpites vão de 1 a 100, a maior diferença possível é 58 (100-42).
# Usamos 60 para uma escala mais arredondada para 10 estrelas.
max_possible_diff_for_bar = 60

print("Início do Jogo da Adivinhação!")

for palpite in palpites:
    tentativas += 1
    diferenca_abs = abs(palpite - secret_number)

    feedback = ""
    if diferenca_abs == 0:
        feedback = "Acertou!"
        acertou = True
    elif diferenca_abs <= 2: # Diferença de 1 ou 2
        feedback = "Muito próximo!"
    elif diferenca_abs <= 5: # Diferença de 3 a 5
        feedback = "Próximo!"
    elif palpite < secret_number: # Palpite abaixo do número secreto
        if diferenca_abs <= 20: # Diferença de 6 a 20 abaixo
            feedback = "Baixo!"
        else: # Diferença maior que 20 abaixo
            feedback = "Muito baixo!"
    else: # palpite > secret_number (Palpite acima do número secreto)
        if diferenca_abs <= 20: # Diferença de 6 a 20 acima
            feedback = "Alto!"
        else: # Diferença maior que 20 acima
            feedback = "Muito alto!"

    # Lógica para a barra de progresso (mais estrelas = mais perto)
    num_stars = 0
    if max_possible_diff_for_bar > 0: # Evita divisão por zero
        # Calcula a pontuação de proximidade: quanto mais perto, maior a pontuação.
        # Max_possible_diff_for_bar - diferenca_abs fará com que 0 diferença resulte em max_possible_diff_for_bar pontos.
        closeness_score = max_possible_diff_for_bar - diferenca_abs
        # Escala a pontuação para 10 estrelas. max(0, ...) garante que não teremos estrelas negativas.
        num_stars = int(max(0, closeness_score / (max_possible_diff_for_bar / 10)))
    elif diferenca_abs == 0: # Caso especial se max_possible_diff_for_bar fosse 0 (nunca seria aqui, mas por robustez)
        num_stars = 10

    progress_bar = "*" * num_stars + "-" * (10 - num_stars) # Cria a string da barra

    print(f"\nPalpite {tentativas}: {palpite}")
    print(f"Feedback: {feedback}")
    print(f"Progresso: [{progress_bar}]")

    if acertou:
        print(f"\nParabéns! Você acertou em {tentativas} tentativas.")
        break

if not acertou:
    print("\nFim dos palpites. O jogador não acertou o número secreto.")
```
