IMC

```python
peso= float(input("Digite seu peso: "))
altura= float(input("Digite sua altura: "))
imc= (altura *altura)/peso
if imc < 18.5:
  print("Abaixo do peso")
elif imc>=18.5 and imc<=24.9:
  print("Peso normal")
elif imc >= 25 and imc<=29.9:
  print("Sobrepeso")
else:
  print("Obesidade")
```

Tarifa de energia

```python
consumo= float(input("Digite o consumo em kWh: "))
if consumo <=100:
  calculo= consumo * 0.5
elif consumo >100 and consumo <=300:
  calculo=consumo*0.75
else:
  calculo=consumo*1.20
print(f"Seu consumo foi: {calculo}")
```

identificar maior ou menor

```python
numero1 = float(input("Digite o primeiro número: "))
numero2 = float(input("Digite o segundo número: "))
numero3 = float(input("Digite o terceiro número: "))

# Encontrando o maior número
maior = numero1
if numero2 > maior:
  maior = numero2
if numero3 > maior:
  maior = numero3

# Encontrando o menor número
menor = numero1
if numero2 < menor:
  menor = numero2
if numero3 < menor:
  menor = numero3

print(f"O maior número é: {maior}")
print(f"O menor número é: {menor}")
```

Soma dos números pares

```python
soma=0
for i in range (1,51): # Changed range to 51 to include 50
 if i % 2 == 0:
   soma= soma + i

# Print the final sum outside the loop
print(f"A soma dos números pares até 50 é: {soma}")
```

caixa eletronico

```python
contador = 0
valor= int(input("Insira o valor que deseja sacar: "))
while valor % 10 !=0:
   print("O valor deve ser multiplo de 10")
   contador += 1
   valor= int(input("Insira o valor que deseja sacar: "))

print("Saque realizado!")
```

Verificação de prefixo e sufixo

```python
string="programação_python.py"
if string.startswith("prog"):
  print("o texto começa com prog")
else:
  print("o texto não começa com prog")
if string.endswith(".py"):
  print("o texto termina com .py")
else:
  print("o texto não termina com .py")
```
