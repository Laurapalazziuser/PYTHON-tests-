exercício 1- calculo de média

```python
def calcular_media(nota1, nota2, nota3):
  media= (nota1 + nota2 + nota3)/3
  return media
media= calcular_media(3,7,8)
if media >=7:
  situacao="aprovado"
else:
  situacao="reprovado"
print("a média é:", media, "e a situação é: ", situacao )
```

exercício 2- verificação número primo

```python
def identificar_primo(numero):
  contador=0
  for i in range(1, numero + 1):
      if numero % i == 0:
         contador +=1
  if contador == 2:
    print(numero, "é primo")
  else:
    print(numero, "não é primo")


numero=int(input("digite um valor: "))
identificar_primo(numero)
```
