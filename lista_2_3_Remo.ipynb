Exercício 4- Contando letras em uma string

```python
import string  #precisa para o programa em Python usar e buscar (necessário para outra linha)
palavra=str(input("Digite uma palavra ou frase: "))
print(sum(palavra.count(letra)for letra in string.ascii_lowercase)) # sum é para somar em uma sequencia; para cada letra no alfabeto,conta quantas vezes aparece na "palavra"; a outra parte "for..." faz parte do programa de python que importei que já vem equipado com as letras minúsculas
```

```python
frase= 'estudando Python na faculdade'
palavra= str(input("digite a palavra que está procurando: "))
indice= frase.find(palavra) #find trabalha com indices e posições
if indice != -1: #0 já é valido, por isso usei um número que não será possível
  print(f"Palavra encontrada no índice: {indice}")
else:
  print("Palavra não encontrada!")
```

Exercício 6- Substituindo palavras

```python
frase= 'Hoje o dia está chuvoso'
palavra= str(input("Digite qual palavra quer trocar: "))
nova_palavra= str(input("Digite a nova palavra: "))
frase_nova= frase.replace(palavra, nova_palavra)
print(frase_nova)
```

Exercício 10- Verificando o Tamanho do Texto


```python
palavra= str(input("Digite uma palavra ou frase: "))
tamanho= len(palavra)
if tamanho < 5:
  print(f"Sua palavra é curta, possui {tamanho} letras")
else:
  print(f"Sua palavra é longa, possui {tamanho} letras")
```
